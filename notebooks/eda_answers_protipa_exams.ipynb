```python
import json
import pandas as pd
from pathlib import Path
import itables
from itables import show
```

```python
#file_path=Path("../results/krikri_test/llama-krikri-8b-instruct-v1.5")
file_path=Path("../results/closed_aggregate_private/llama-krikri-8b-instruct-v1.5")
```

```python
files = file_path.glob("*.jsonl")
print (files)
```

```python
parsed_data = []

for file in  files:
    with open (file, "r", encoding="utf-8") as f:
        for line in f:
            item = json.loads(line)
            row = {
            "id": item["doc"].get("id"),
            "subject": item["doc"].get("subject"),
            "format": item["doc"].get("format"),
            "input": item["doc"].get("input"),
            "question": item["doc"].get("question"),
            "choices": item["doc"].get("choices"),
            "target": item.get("target"),
            "filtered_resps": item.get("filtered_resps", [""])[0] if item.get("filtered_resps") else "",
            "bertscore": item.get("bertscore_f1_max", None),
            "exact_match": item.get("exact_match", None)
            }
            parsed_data.append(row)
```

```python
df = pd.DataFrame (parsed_data)
pd.set_option('display.max_columns', None)
df
```

```python
show(df,
     layout={"top1": "searchBuilder"},
     buttons=[
         "pageLength",
         {"extend": "excelHtml5", "title": "eda_answers_protipa_exams"}
     ]
)
```
